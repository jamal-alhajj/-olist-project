```python
import pandas as pd


RAW_PATH = "../data/raw/"


customers = pd.read_csv(RAW_PATH + "olist_customers_dataset.csv")
orders = pd.read_csv(RAW_PATH + "olist_orders_dataset.csv")
items = pd.read_csv(RAW_PATH + "olist_order_items_dataset.csv")
products = pd.read_csv(RAW_PATH + "olist_products_dataset.csv")


def explore_df(df, name):
    print(f"--- {name} ---")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    print("Data types:\n", df.dtypes)
    print("Missing values:\n", df.isnull().sum())
    print("First 5 rows:\n", df.head(), "\n\n")

# Explore all datasets
explore_df(customers, "Customers")
explore_df(orders, "Orders")
explore_df(items, "Order Items")
explore_df(products, "Products")
```
